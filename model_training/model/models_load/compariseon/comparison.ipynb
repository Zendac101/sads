```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# 1. Define Model Performance Data
models = ['SVM', 'Random Forest', 'LSTM','GXboost']

regression_metrics = {
    'MAE': [3.5533, 2.3904, 1.5000,2.0731],
    'MSE': [26.4663, 16.0666, 6.1100,12.1586],
    'RMSE': [5.1445, 4.0083, 2.4700,3.4869],
    'R2 Score': [0.8833, 0.9292, 0.9739,0.9483]
}

classification_metrics = {
    'Accuracy': [0.9360, 0.9460, 0.9687,0.9449],
    'Precision': [0.9389, 0.9445, 0.9684,0.9435],
    'Recall': [0.9360, 0.9460, 0.9687,0.9449],
    'F1-Score': [0.9371, 0.9440, 0.9684,0.9483]
}

# Set consistent colors for the models
colors = ["#3f6094", "#48975b", "#963639","#943f86"]


# ==========================================
# PLOT 1: Regression Metrics Comparison
# ==========================================
fig, axs = plt.subplots(2, 2, figsize=(12, 10))
axs = axs.ravel()

for i, (metric, values) in enumerate(regression_metrics.items()):
    bars = axs[i].bar(models, values, color=colors, edgecolor='black', width=0.5)
    # Using raw string or escaping for LaTeX presentation if required
    title_label = r'$R^2$ Score' if metric == 'R2 Score' else metric
    axs[i].set_title(f'Model Comparison: {title_label}', fontsize=14, fontweight='bold')
    axs[i].set_ylabel('Value', fontsize=12)
    axs[i].tick_params(axis='both', which='major', labelsize=11)
    axs[i].grid(axis='y', linestyle='--', alpha=0.5)
    
    # Add precise values on top of the bars
    for bar in bars:
        height = bar.get_height()
        axs[i].annotate(f'{height:.4f}',
                        xy=(bar.get_x() + bar.get_width() / 2, height),
                        xytext=(0, 3),  # 3 points vertical offset
                        textcoords="offset points",
                        ha='center', va='bottom', fontsize=10, fontweight='bold')

plt.tight_layout()
plt.savefig('regression_metrics_comparison.png', dpi=300)
plt.close()


# ==========================================
# PLOT 2: Classification Metrics Subplots
# ==========================================
fig, axs = plt.subplots(2, 2, figsize=(12, 10))
axs = axs.ravel()

for i, (metric, values) in enumerate(classification_metrics.items()):
    bars = axs[i].bar(models, values, color=colors, edgecolor='black', width=0.5)
    axs[i].set_title(f'Model Comparison: {metric}', fontsize=14, fontweight='bold')
    axs[i].set_ylabel('Score', fontsize=12)
    axs[i].set_ylim(0.85, 1.0)  # Magnify top region to clearly see performance delta
    axs[i].tick_params(axis='both', which='major', labelsize=11)
    axs[i].grid(axis='y', linestyle='--', alpha=0.5)
    
    # Add precise values on top of the bars
    for bar in bars:
        height = bar.get_height()
        axs[i].annotate(f'{height:.4f}',
                        xy=(bar.get_x() + bar.get_width() / 2, height),
                        xytext=(0, 3),  # 3 points vertical offset
                        textcoords="offset points",
                        ha='center', va='bottom', fontsize=10, fontweight='bold')

plt.tight_layout()
plt.savefig('classification_metrics_comparison.png', dpi=300)
plt.close()


# ==========================================
# PLOT 3: Consolidated Grouped Bar Chart
# ==========================================
df_clf = pd.DataFrame(classification_metrics, index=models)
fig, ax = plt.subplots(figsize=(11, 6))

# Plot using pandas built-in bar tool mapped onto the explicit axis
df_clf.plot(kind='bar', ax=ax, edgecolor='black', width=0.8)

ax.set_title('Overall Classification Metrics Side-by-Side', fontsize=16, fontweight='bold')
ax.set_ylabel('Score', fontsize=12)
ax.set_xlabel('Model', fontsize=12)
plt.xticks(rotation=0, fontsize=11)
ax.set_ylim(0.90, 1.0)
ax.legend(title='Metrics', fontsize=10, title_fontsize=11, loc='lower right')
ax.grid(axis='y', linestyle='--', alpha=0.7)

# Add accurate text labels to each bar group
for p in ax.patches:
    ax.annotate(f'{p.get_height():.3f}', 
                (p.get_x() + p.get_width() / 2., p.get_height()), 
                ha='center', va='center', xytext=(0, 8), 
                textcoords='offset points', fontsize=9, fontweight='bold')

plt.tight_layout()
plt.savefig('consolidated_classification_comparison.png', dpi=300)
plt.close()

print("All plots generated and saved as high-resolution PNGs successfully.")
```

```python
import numpy as np
import pandas as pd
import joblib
from tensorflow.keras.models import load_model, Model
```

```python
# 1. Load the components
# Note: You would need to have saved the scaler and feature_extractor previously
svm_model = joblib.load('svm_model.pkl')
scaler = joblib.load('scaler.pkl') 
feature_extractor = joblib.load('feature_extractor.h5')
```

```python
# 2. Prepare sample input data (matching the features in your training set)

# Updated Step 2 & 3
features = ['aqi', 'so2', 'co', 'o3', 'pm10', 'pm2.5', 'no2', 'nox', 'no']
# Create a DataFrame so the scaler recognizes the feature names
sample_df = pd.DataFrame([[80.0, 0.9, 0.17, 35.0, 18.0, 17.0, 12.3, 12.6, 0.3]], 
                         columns=features)
```

```python
# 3. Scale the input
sample_scaled = scaler.transform(sample_df)

# 4. Extract features using the Neural Network layer
sample_features = feature_extractor.predict(sample_scaled)
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
le.classes_ = np.array([ 'Good', 'Moderate', 'Unhealthy for Sensitive Groups'])
# 5. Predict using the SVM
prediction_index = svm_model.predict(sample_features)


# 6. Decode the result back to the status string
# (Uses the classes defined in your LabelEncoder: 'Good', 'Moderate', etc.)
predicted_status = le.inverse_transform(prediction_index)
```

```python
print(f"Predicted Air Quality Status: {predicted_status[0]}")
```
